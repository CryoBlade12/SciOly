````python
class EventTimeSlots:
    """
    Maps each event to its conflict block group based on official Science Olympiad
    National Tournament schedule releases. Events with the same integer value (1-6)
    share the same time slot and will conflict.

    Group 7 represents self-scheduled / impound build events.
    """

    DIV_B_BLOCKS = {
        # Block 1
        "codebusters": 1,
        "disease detectives": 1,
        "remote sensing": 1,

        # Block 2
        "water quality": 2,
        "experimental design": 2,
        "solar system": 2,

        # Block 3
        "botany": 3,
        "meteorology": 3,

        # Block 4
        "circuit lab": 4,
        "dynamic planet": 4,
        "food science": 4,

        # Block 5
        "heredity": 5,
        "rocks and minerals": 5,
        "thermodynamics": 5,

        # Block 6
        "anatomy and physiology": 6,
        "crime busters": 6,
        "write it do it": 6,

        # Block 7 (Self-Scheduled / Build Events)
        "boomilever": 7,
        "elastic launched glider": 7,
        "hovercraft": 7,
        "ping pong parachute": 7,
        "roller coaster": 7,
        "scrambler": 7
    }

    DIV_C_BLOCKS = {
        # Block 1
        "anatomy and physiology": 1,
        "engineering cad": 1,
        "forensics": 1,

        # Block 2
        "codebusters": 2,
        "disease detectives": 2,
        "remote sensing": 2,
        "materials science": 2,

        # Block 3
        "astronomy": 3,
        "botany": 3,
        "experimental design": 3,

        # Block 4
        "chemistry lab": 4,
        "rocks and minerals": 4,
        "entomology": 4,

        # Block 5
        "circuit lab": 5,
        "protein modeling": 5,
        "water quality": 5,

        # Block 6
        "designer genes": 6,
        "dynamic planet": 6,
        "thermodynamics": 6,

        # Block 7 (Self-Scheduled / Build Events)
        "boomilever": 7,
        "electric vehicle": 7,
        "hovercraft": 7,
        "mission possible": 7,
        "ping pong parachute": 7,
        "wright stuff": 7,
        "bungee drop": 7
        , "helicopter": 7
        , "machines": 7
        , "robot tour": 7
    }

    @classmethod
    def get_block(cls, event_name, division='C'):
        """
        Retrieves the conflict block number for a given event.

        Args:
            event_name (str): The name of the event.
            division (str): 'B' for middle school, 'C' for high school.

        Returns:
            int: The block number (1-7), or None if the event is not found.
        """
        event = event_name.lower().strip()
        if division.upper() == 'C':
            return cls.DIV_C_BLOCKS.get(event, None)
        elif division.upper() == 'B':
            return cls.DIV_B_BLOCKS.get(event, None)

        return None

    @classmethod
    def get_conflicts(cls, event_name, division='C'):
        """
        Returns a list of all other events that conflict with the given event.
        """
        block = cls.get_block(event_name, division)
        if block is None or block == 7: # Group 7 events don't strictly conflict
            return []

        if division.upper() == 'C':
            schedule = cls.DIV_C_BLOCKS
        else:
            schedule = cls.DIV_B_BLOCKS

        return [e for e, b in schedule.items() if b == block and e != event_name.lower().strip()]

# --- Example Usage ---
'''
if __name__ == "__main__":
    my_event = "disease detectives"

    # Find out what time block it is in for Division C
    block = EventTimeSlots.get_block(my_event, 'C')
    print(f"'{my_event.title()}' is in Block {block}")

    # Find what other events run at the exact same time
    conflicts = EventTimeSlots.get_conflicts(my_event, 'C')
    print(f"Conflicts: {', '.join(e.title() for e in conflicts)}")
'''
````

#Retrieve Current Events

````python
import requests
from bs4 import BeautifulSoup

def get_current_events(year):
    """
    Scrapes the Science Olympiad Wiki to obtain the actual events for a given year.
    Falls back to EVENTS_MAP keys if the year page is not found.
    """
    # Science Olympiad Wiki usually follows this URL pattern for yearly slates
    url = f"https://scioly.org/wiki/index.php/{year}"

    try:
        response = requests.get(url, timeout=10)
        if response.status_code != 200:
            print(f"Warning: Could not find page for {year}. Falling back to default map.")
            return list(EVENTS_MAP.keys())

        soup = BeautifulSoup(response.text, 'html.parser')

        # Locate the events table or list (typically under 'Events' header)
        current_events = []

        # Find the header/section specifically for Division C using the title attribute
        div_c_header = soup.find(lambda tag: tag.get('title') == "Division C")

        if div_c_header:
            # Update: parent is now the specific table containing the header
            parent = div_c_header.find_parent('table')
            if parent:
                links = parent.find_all('a')
                for link in links:
                    event_candidate = link.text.strip().lower()
                    if "division c" in event_candidate:
                        continue
                    current_events.append(event_candidate)
                    if event_candidate not in EVENTS_MAP:
                        print(f"Warning: '{event_candidate}' not in EVENTS_MAP")

        # Deduplicate
        current_events = list(dict.fromkeys(current_events))

        if not current_events:
            print(f"Warning: No Division C events found for {year}.")
            return list(EVENTS_MAP.keys())

        return current_events

    except Exception as e:
        print(f"Error fetching events: {e}")
        return list(EVENTS_MAP.keys())
````

#Generated Code

````python
"""Build Science Olympiad teams A/B/C and staff every event, by exact optimisation.

The whole thing is expressed as three integer programs, solved in sequence A, then
B, then C -- which is exactly what criterion 8 asks for: "maximize the score
pairings for A, with B having the highest scores after A has been chosen, and C
having the remaining lower scored people".

Each stage answers one question: **of everyone still unplaced, which team_size
people should form this team, and which events should each of them run, so that
this team's total score/preference value is as high as possible?** Because team
membership and event staffing are decided **together** rather than one after the
other, team A is not merely the 15 strongest individuals -- it is the 15 who best
*cover the events*, which is a materially different and better answer.

Why an ILP rather than a greedy pass: every constraint here (exact event
capacities, at most one event per time block, at most three events per person,
at most seven 12th graders per team) interacts with the others, and a greedy
assignment that respects them one at a time is easy to write and very easy to get
subtly wrong. CBC returns a provably optimal solution or tells us the instance is
infeasible; there is no third outcome where it quietly returns something wrong.

Modelling notes
---------------
* **EVENT_CAPACITY is read as **per team**: each of A, B and C fields exactly
  ``EVENT_CAPACITY[event]`` people in every event, which is how a Science
  Olympiad school actually enters three teams. Criterion 8 assumes this too --
  it wants the best people in an event on A, the next best on B, the rest on C,
  which only makes sense if all three teams enter that event.
* **Time blocks** conflict by equality: two events collide exactly when they map
  to the same block, so a person may hold at most one event per block.
* **Grade level is not experience.** Grade level is used only for the 12th
  grade cap and never ranks anybody; a 12th grader in their first season and
  one in their fourth are treated very differently. Years of Science Olympiad
  are a separate input, and they influence *which team* a person lands on
  rather than which event they run -- seasons of experience say nothing about
  whether somebody should be in Astronomy or Forensics, which is what the
  per-event scores are for.
* **Team C is the overflow team and is exempt from the senior cap.** A and B
  take at most ``max_seniors_per_team`` twelfth graders each; C takes whoever is
  left, however many of them are seniors. Capping C as well would put a hard
  ceiling on how many seniors the school could enter at all.
"""

from __future__ import annotations

import warnings
from dataclasses import dataclass, field
from typing import Any, Iterable, Mapping, NamedTuple, Sequence

import pulp

TEAMS = ("A", "B", "C")


class Weights(NamedTuple):
    """How much each signal counts when building one team.

    ``score`` and ``preference`` apply per **event assignment**, because both are
    about whether this person suits this event. ``experience`` applies per **team
    place**, because seasons of Science Olympiad say nothing about which event
    somebody should run -- it is a statement about the person.
    """

    score: float
    preference: float
    experience: float


# Criterion 9, plus experience:
#   A -- score is what matters, veterans favoured, preference only breaks near-ties
#   B -- an even mix of score and preference, experience still counts
#   C -- preference dominant; the newer members mostly land here
#
# All three are denominated in the same unit: the worth of one event pairing. So
# `experience=0.25` means "a maximally seasoned competitor is worth a quarter of an
# event pairing", which keeps experience a strong tiebreak without letting it
# overturn a clear difference in demonstrated score.
TEAM_WEIGHTS: dict[str, Weights] = {
    "A": Weights(score=1.00, preference=0.10, experience=0.25),
    "B": Weights(score=0.50, preference=0.50, experience=0.20),
    "C": Weights(score=0.15, preference=0.85, experience=0.10),
}

# A person with a real score in an event always beats one with no record at all,
# so measured scores are mapped into [SCORE_FLOOR, 1] and "no score" is 0.
SCORE_FLOOR = 0.10

# Starting points for :func:`build_score_buckets`. Both are calibration dials
# rather than settled facts -- see that function's docstring.
MEDAL_POINTS = {1: 10.0, 2: 7.0, 3: 5.0, 4: 3.0, 5: 2.0}
DEFAULT_COURSE_POINTS = 2.0

SENIOR_GRADE = 12
POSSIBLE_GRADE_LEVELS = (9, 10, 11, 12)

# Most seasons somebody in each grade could plausibly have behind them: a 9th
# grader can only have competed in middle school, so two is their ceiling, and
# each later grade adds one. Used to reject impossible data -- a 9th grader
# recorded with four seasons is a typo, not a prodigy.
MAX_EXPERIENCE_BY_GRADE = {9: 2, 10: 3, 11: 4, 12: 5}

# The last team absorbs everyone left over, so the senior cap does not apply to it.
OVERFLOW_TEAM = TEAMS[-1]


def _senior_caps(sizes: Mapping[str, int], max_seniors_per_team: int) -> dict[str, int]:
    """How many 12th graders each team may hold.

    The overflow team's cap is its own size, which is no cap at all -- expressing
    "exempt" as a non-binding number keeps every downstream calculation uniform
    instead of special-casing one team everywhere.
    """
    return {
        team: sizes[team] if team == OVERFLOW_TEAM else max_seniors_per_team
        for team in TEAMS
    }


def default_solver() -> Any:
    """The best MILP solver available, quiet.

    Prefers ``COIN_CMD``; falls back to the CBC binary bundled with pulp, whose
    constructor is deprecated but is the only solver present in a plain install.
    """
    for name in ("COIN_CMD", "PULP_CBC_CMD"):
        factory = getattr(pulp, name, None)
        if factory is None:
            continue
        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", DeprecationWarning)
                solver = factory(msg=0)
            if solver.available():
                return solver
        except Exception:  # solver not installed on this machine
            continue
    raise RuntimeError(
        "no MILP solver available; install one with `pip install pulp[cbc]`"
    )


def _binary(problem: pulp.LpProblem, name: str) -> pulp.LpVariable:
    """A binary variable attached to ``problem``, across pulp 3.x and 4.x."""
    if hasattr(problem, "add_variable"):
        return problem.add_variable(name, 0, 1, cat="Binary")
    return pulp.LpVariable(name, cat="Binary")


def _seniors_required(
    pool: Sequence[str],
    size: int,
    later: Sequence[tuple[int, int]],
    grade_levels: Mapping[str, int],
) -> int:
    """Fewest 12th graders this team must take so the later teams stay feasible.

    Solving A, then B, then C in sequence has a trap: nothing stops an earlier
    team from filling up with underclassmen when those happen to score best,
    stranding seniors on teams that have no room for them. So each team is also
    told the minimum number of seniors it has to absorb, from two directions:

    * the teams after this one can hold only so many seniors between them, so any
      surplus beyond that must be taken now;
    * those teams also need enough underclassmen to fill their own non-senior
      places, so this team cannot take more juniors than are spare.

    Args:
        later: ``(size, senior_cap)`` for each team still to be built. The
            overflow team's cap equals its size, so it soaks up any surplus and
            demands no underclassmen -- which is what makes both bounds vanish.
    """
    seniors = sum(1 for p in pool if grade_levels[p] == SENIOR_GRADE)
    juniors = len(pool) - seniors

    surplus = seniors - sum(cap for _, cap in later)
    juniors_needed_later = sum(max(0, later_size - cap) for later_size, cap in later)
    juniors_spare = juniors - juniors_needed_later
    return max(0, surplus, size - juniors_spare)


@dataclass
class Roster:
    """Result of :func:`build_teams`."""

    # person -> team
    team_of: dict[str, str] = field(default_factory=dict)
    # team -> members
    teams: dict[str, list[str]] = field(default_factory=dict)
    # team -> (event -> people from that team competing in it)
    event_assignments: dict[str, dict[str, list[str]]] = field(default_factory=dict)
    # person -> the events they compete in
    person_events: dict[str, list[str]] = field(default_factory=dict)
    # team -> the optimal objective value the solver achieved
    objective: dict[str, float] = field(default_factory=dict)
    # person -> (event -> normalised score in [0, 1]), the values ranked on
    score_index: dict[str, dict[str, float]] = field(default_factory=dict)
    # person -> normalised years of Science Olympiad in [0, 1]
    experience_index: dict[str, float] = field(default_factory=dict)
    # team members who compete in nothing, because their team has more people
    # than the schedule has places. Only ever the overflow team.
    reserves: list[str] = field(default_factory=list)

    def preferences_granted(self, preferences: Mapping[str, Sequence[str]]) -> int:
        """How many assignments matched something the person actually asked for."""
        return sum(
            1
            for person, events in self.person_events.items()
            for event in events
            if event in preferences.get(person, ())
        )


def _field(entry: Any, name: str) -> Any:
    """Read ``name`` off a mapping, dataclass or namedtuple entry."""
    if isinstance(entry, Mapping):
        return entry.get(name)
    return getattr(entry, name, None)


def _people(
    grade_levels: Mapping[str, int],
    preferences: Mapping[str, Sequence[str]],
    scores: Mapping[str, Mapping[str, float]],
    years_experience: Mapping[str, float],
) -> list[str]:
    """Everyone named in any input, in a stable order."""
    return sorted(
        set(grade_levels) | set(preferences) | set(scores) | set(years_experience)
    )


def normalised_scores(
    people: Sequence[str],
    events: Sequence[str],
    scores: Mapping[str, Mapping[str, float]],
) -> dict[str, dict[str, float]]:
    """Rescale each event's raw scores onto a common [0, 1] axis.

    Events are normalised independently, so a subject marked out of 30 and one
    marked out of 100 carry equal weight. Anyone with a recorded score lands in
    ``[SCORE_FLOOR, 1]``; anyone with no record in that event scores 0, so
    measured evidence always outranks its absence.
    """
    index: dict[str, dict[str, float]] = {person: {} for person in people}
    for event in events:
        recorded = {
            person: float(scores[person][event])
            for person in people
            if event in scores.get(person, {})
        }

        if not recorded:
            continue

        low, high = min(recorded.values()), max(recorded.values())
        span = high - low

        for person, value in recorded.items():
            fraction = 1.0 if span == 0 else (value - low) / span
            index[person][event] = SCORE_FLOOR + (1 - SCORE_FLOOR) * fraction

    return index


def build_score_buckets(
    medals: Mapping[str, Sequence[Any]],
    coursework: Mapping[str, Sequence[str]],
    course_events: Mapping[str, Any],
    medal_points: Mapping[int, float] = MEDAL_POINTS,
    course_points: float = DEFAULT_COURSE_POINTS,
    medals_outrank_coursework: bool = True,
) -> dict[str, dict[str, float]]:
    """Accumulate per-event point buckets from results and relevant coursework.

    Two things feed a bucket: medalling in that event, and studying something that
    prepares you for it (AP Chemistry pours into Chemistry Lab, Materials Science
    and Forensics). The output is the ``scores`` mapping :func:`build_teams` wants.

    **Having done it beats having studied it.** By default the two are not simply
    added up, because addition lets a stack of loosely related classes out-total a
    medal, and somebody who has actually competed is better evidence than somebody
    who has only sat the course. Coursework is instead compressed into a band
    narrower than the closest pair of medals, so it orders people who have never
    medalled in an event and breaks ties between people who medalled equally, but
    can never lift anyone above a better result. Pass
    ``medals_outrank_coursework=False`` for the plain additive behaviour.

    **What is left to calibrate.** In the default mode ``course_points`` scales
    every course alike and is then normalised away, so the dial that still bites
    is the *relevance* map in ``course_events`` -- how strongly each class feeds
    each event. In additive mode ``course_points`` becomes a real dial again, and
    only its ratio to ``medal_points`` matters, since :func:`normalised_scores`
    rescales each event independently afterwards.

    Args:
        medals: person -> results, each a mapping (or object) with ``"event"`` and
            ``"place"``. Places absent from ``medal_points`` score nothing.
        coursework: person -> the courses they have taken.
        course_events: course -> either the events it feeds, or a mapping of
            event to how strongly it feeds it, letting AP Chemistry count fully
            towards Chemistry Lab but only partly towards Forensics.
        medal_points: placing -> points awarded. Whole numbers keep the guarantee
            above exact; the band is derived from the closest pair among them.
        course_points: points a fully relevant course contributes.
        medals_outrank_coursework: keep any medal ahead of any amount of study.

    Returns:
        person -> (event -> points). Events nobody has points in are absent
        rather than zero, so "no evidence" stays distinct from "scored nothing".
    """
    people = sorted(set(medals) | set(coursework))
    from_medals: dict[str, dict[str, float]] = {person: {} for person in people}
    from_courses: dict[str, dict[str, float]] = {person: {} for person in people}

    for person in people:
        for result in medals.get(person, ()):
            event = _field(result, "event")
            place = _field(result, "place")
            if event is None or place is None:
                raise ValueError(
                    f"medals[{person!r}] entries need an 'event' and a 'place'; "
                    f"got {result!r}"
                )
            points = medal_points.get(place)
            if points:
                tally = from_medals[person]
                tally[event] = tally.get(event, 0.0) + points

        for course in coursework.get(person, ()):
            feeds = course_events.get(course)
            if feeds is None:
                raise ValueError(
                    f"course {course!r} taken by {person} is not in course_events; "
                    f"add the events it prepares for, or drop it from coursework"
                )
            relevance = (
                feeds if isinstance(feeds, Mapping) else dict.fromkeys(feeds, 1.0)
            )
            for event, weight in relevance.items():
                tally = from_courses[person]
                tally[event] = tally.get(event, 0.0) + course_points * weight

    scale = 1.0
    if medals_outrank_coursework:
        ceiling = max(
            (points for tally in from_courses.values() for points in tally.values()),
            default=0.0,
        )
        scale = 0.0 if ceiling == 0 else _coursework_band(medal_points) / ceiling

    buckets: dict[str, dict[str, float]] = {}
    for person in people:
        combined: dict[str, float] = {}
        for event, points in from_medals[person].items():
            combined[event] = combined.get(event, 0.0) + points
        for event, points in from_courses[person].items():
            combined[event] = combined.get(event, 0.0) + points * scale
        buckets[person] = {
            event: points for event, points in combined.items() if points > 0
        }

    return buckets


def _coursework_band(medal_points: Mapping[int, float]) -> float:
    """Widest coursework may spread without ever overtaking a better result.

    Kept just inside both the lowest medal (so study alone never passes a medal)
    and the closest gap between two medals (so it never reorders two people who
    both medalled).
    """
    distinct = sorted(set(medal_points.values()))
    if not distinct:
        return 1.0  # no medal table at all: coursework is the only signal
    gaps = [later - earlier for earlier, later in zip(distinct, distinct[1:])]
    return 0.9 * min([distinct[0]] + gaps)


def normalised_experience(
    people: Sequence[str],
    years_experience: Mapping[str, float],
) -> dict[str, float]:
    """Rescale years of Science Olympiad onto [0, 1] across the whole squad.

    Relative, not absolute: in a veteran squad four years is unremarkable, in a
    young one it is the most experience anyone has. Where everybody has served
    the same length of time the term is constant and therefore cannot influence
    the choice at all, which is the right behaviour.
    """
    values = [float(years_experience.get(person, 0)) for person in people]
    low, high = min(values), max(values)
    span = high - low
    if span == 0:
        return {person: 1.0 for person in people}
    return {
        person: (float(years_experience.get(person, 0)) - low) / span
        for person in people
    }


def preference_index(
    people: Sequence[str],
    preferences: Mapping[str, Sequence[str]],
) -> dict[str, dict[str, float]]:
    """Turn each person's wish list into weights.

    The list is a **set**, not a rankings: people name five events they would like
    with no order between them, so every listed event scores 1.0 and the position
    in the list carries no meaning. Anything unlisted simply does not appear.
    """
    return {
        person: dict.fromkeys(preferences.get(person, ()), 1.0) for person in people
    }


def _validate(
    people: Sequence[str],
    events: Sequence[str],
    grade_levels: Mapping[str, int],
    years_experience: Mapping[str, float],
    max_experience_by_grade: Mapping[int, int],
    event_time_blocks: Mapping[str, Any],
    event_capacity: Mapping[str, int],
    sizes: Mapping[str, int],
    senior_caps: Mapping[str, int],
    max_events_per_person: int,
    everyone_competes: bool,
) -> None:
    """Reject impossible inputs up front, with the arithmetic that proves it."""
    if not people:
        raise ValueError("no people to place")
    if not events:
        raise ValueError("no events to staff")
    if len(set(events)) != len(events):
        duplicates = sorted(e for e in events if list(events).count(e) > 1)
        raise ValueError(f"events must be unique; duplicated: {duplicates}")

    missing_block = [e for e in events if e not in event_time_blocks]
    if missing_block:
        raise ValueError(f"no time block given for: {sorted(missing_block)}")
    missing_capacity = [e for e in events if e not in event_capacity]
    if missing_capacity:
        raise ValueError(f"no EVENT_CAPACITY given for: {sorted(missing_capacity)}")

    non_positive = sorted(e for e in events if event_capacity[e] < 1)
    if non_positive:
        raise ValueError(
            "EVENT_CAPACITY must be at least 1 so every event is staffed; "
            f"got 0 or less for: {non_positive}"
        )

    bad_grades = sorted(
        p for p in people if grade_levels.get(p) not in POSSIBLE_GRADE_LEVELS
    )

    if bad_grades:
        raise ValueError(
            f"every person needs a grade level in {POSSIBLE_GRADE_LEVELS}; missing or "
            f"invalid for: {bad_grades}"
        )

    # Required rather than defaulted to zero: a mistyped name would otherwise
    # silently become a first-year and be quietly sorted down the teams.
    bad_experience = sorted(
        p
        for p in people
        if not isinstance(years_experience.get(p), (int, float))
        or isinstance(years_experience.get(p), bool)
        or years_experience[p] < 0
    )
    if bad_experience:
        raise ValueError(
            "every person needs years_experience as a number of seasons (0 or "
            f"more); missing or invalid for: {bad_experience}"
        )

    impossible = [
        (p, grade_levels[p], years_experience[p])
        for p in people
        if years_experience[p] > max_experience_by_grade[grade_levels[p]]
    ]
    if impossible:
        detail = "; ".join(
            f"{p} is in grade {grade} with {years} seasons but grade {grade} "
            f"allows at most {max_experience_by_grade[grade]}"
            for p, grade, years in sorted(impossible)
        )
        raise ValueError(f"years_experience impossible for grade level: {detail}")

    smallest = min(sizes.values())
    if smallest < 1:
        raise ValueError(
            f"not enough people: {len(people)} cannot fill teams of "
            f"{sizes['A']} and {sizes['B']} and still leave anyone for C"
        )

    # A person holds at most one event per block, so a team needs at least as many
    # members as the busiest block demands.
    per_block: dict[Any, int] = {}
    for event in events:
        block = event_time_blocks[event]
        per_block[block] = per_block.get(block, 0) + event_capacity[event]
    busiest_block, busiest_demand = max(per_block.items(), key=lambda kv: (kv[1], str(kv[0])))
    if busiest_demand > smallest:
        raise ValueError(
            f"time block {busiest_block!r} needs {busiest_demand} people at once "
            f"but the smallest team has {smallest}; no team can cover it"
        )

    total_slots = sum(event_capacity[e] for e in events)
    if total_slots > max_events_per_person * smallest:
        raise ValueError(
            f"events need {total_slots} places per team but the smallest team of "
            f"{smallest} can supply only {max_events_per_person * smallest} "
            f"({max_events_per_person} events each)"
        )

    # Only the fixed-size teams have to fit the places. The overflow team is
    # allowed to be bigger than the event schedule can field -- that is what makes
    # it the overflow team -- and its surplus members become reserves.
    fixed = max(sizes[team] for team in TEAMS if team != OVERFLOW_TEAM)
    if everyone_competes and total_slots < fixed:
        raise ValueError(
            f"only {total_slots} places per team, but teams {', '.join(t for t in TEAMS if t != OVERFLOW_TEAM)} "
            f"are fixed at {fixed}, so {fixed - total_slots} of their members "
            "would sit out; add events, raise EVENT_CAPACITY, lower team_size, "
            "or pass everyone_competes=False"
        )

    seniors = sum(1 for p in people if grade_levels[p] == SENIOR_GRADE)
    senior_room = sum(senior_caps.values())
    if seniors > senior_room:
        capped = ", ".join(
            f"{team} ({senior_caps[team]})" for team in TEAMS if team != OVERFLOW_TEAM
        )
        raise ValueError(
            f"{seniors} twelfth graders cannot be placed: teams take at most "
            f"{senior_room} between them ({capped}, and {OVERFLOW_TEAM} limited "
            f"only by its size of {sizes[OVERFLOW_TEAM]})"
        )

    juniors_needed = sum(max(0, sizes[team] - senior_caps[team]) for team in TEAMS)
    juniors = len(people) - seniors
    if juniors < juniors_needed:
        raise ValueError(
            f"only {juniors} people below grade {SENIOR_GRADE}, but the teams need "
            f"{juniors_needed} of them to respect the senior cap"
        )


def _solve_team(
    team: str,
    pool: Sequence[str],
    size: int,
    senior_cap: int,
    later: Sequence[tuple[int, int]],
    events: Sequence[str],
    event_capacity: Mapping[str, int],
    blocks: Mapping[Any, list[str]],
    score_index: Mapping[str, Mapping[str, float]],
    pref_index: Mapping[str, Mapping[str, float]],
    experience: Mapping[str, float],
    grade_levels: Mapping[str, int],
    max_events_per_person: int,
    everyone_competes: bool,
    solver: Any,
) -> tuple[list[str], dict[str, list[str]], float]:
    """Choose this team's members **and** their events in one optimisation.

    Returns ``(members, {event: people}, objective)``.
    """
    weights = TEAM_WEIGHTS[team]
    problem = pulp.LpProblem(f"science_olympiad_team_{team}", pulp.LpMaximize)

    # y[i] -- person i is on this team. x[i, j] -- person i competes in event j.
    # Indices rather than names keep the LP variable names well formed.
    y = [_binary(problem, f"y_{i}") for i in range(len(pool))]
    x = {
        (i, j): _binary(problem, f"x_{i}_{j}")
        for i in range(len(pool))
        for j in range(len(events))
    }

    def pairing_value(i: int, j: int) -> float:
        """Worth of putting person i in event j: how well they suit that event."""
        person, event = pool[i], events[j]
        return (
            weights.score * score_index[person].get(event, 0.0)
            + weights.preference * pref_index[person].get(event, 0.0)
        )

    # Two kinds of value, on the two kinds of variable. Event pairings are scored
    # per assignment; experience is scored once per person, because it is a
    # property of who is on the team rather than of what they are entered into.
    #
    # Experience is scaled by how many events a member carries on average, so the
    # balance between the two does not drift with roster size; without it, a team
    # whose members each run three events would drown the experience term, while
    # one running a single event each would be dominated by it.
    events_per_member = sum(event_capacity[e] for e in events) / size
    problem += pulp.lpSum(
        pairing_value(i, j) * x[i, j] for i, j in x
    ) + weights.experience * events_per_member * pulp.lpSum(
        experience[pool[i]] * y[i] for i in range(len(pool))
    )

    # Criterion 2: exact team size.
    problem += pulp.lpSum(y) == size, "team_size"

    # Criterion 3: at most seven 12th graders, except on the overflow team, whose
    # cap equals its size and so constrains nothing.
    seniors = [i for i, p in enumerate(pool) if grade_levels[p] == SENIOR_GRADE]
    if seniors and senior_cap < size:
        problem += (
            pulp.lpSum(y[i] for i in seniors) <= senior_cap,
            "senior_cap",
        )

    # ...and at least enough of them that the teams after this one stay solvable.
    required = _seniors_required(pool, size, later, grade_levels)
    if required:
        problem += (
            pulp.lpSum(y[i] for i in seniors) >= required,
            "senior_floor",
        )

    # Criteria 6 and 7: every event filled to exactly its capacity.
    for j, event in enumerate(events):
        problem += (
            pulp.lpSum(x[i, j] for i in range(len(pool))) == event_capacity[event],
            f"capacity_{j}",
        )

    # A team can only field as many competitors as there are places. The overflow
    # team may well be bigger than that, so participation is handled in two ways:
    #
    # * team fits the places -- require everybody to compete, otherwise one strong
    #   person covers several events and the spare seats get padded for free with
    #   whoever, quietly defeating criterion 8;
    # * team is larger than the places -- somebody must sit out, so hold everyone
    #   to a single event. That spreads the work over as many members as there are
    #   places, which is the most participation the numbers allow, and leaves the
    #   rest as named reserves rather than pretending they are competing.
    total_slots = sum(event_capacity[event] for event in events)
    oversubscribed = size > total_slots
    must_all_compete = everyone_competes and not oversubscribed
    per_person_cap = 1 if (everyone_competes and oversubscribed) else max_events_per_person

    for i in range(len(pool)):
        # Criterion 5: only team members compete, and at most three events each.
        for j in range(len(events)):
            problem += x[i, j] <= y[i], f"member_{i}_{j}"
        entered = pulp.lpSum(x[i, j] for j in range(len(events)))
        problem += entered <= per_person_cap, f"max_events_{i}"
        if must_all_compete:
            problem += entered >= y[i], f"active_{i}"
        # Criterion 4: at most one event per time block.
        for block, block_events in blocks.items():
            if len(block_events) < 2:
                continue  # a lone event in a block cannot clash with anything
            problem += (
                pulp.lpSum(x[i, events.index(e)] for e in block_events) <= 1,
                f"block_{i}_{block}",
            )

    problem.solve(solver)
    if pulp.LpStatus[problem.status] != "Optimal":
        raise ValueError(
            f"no valid roster exists for team {team}: solver returned "
            f"{pulp.LpStatus[problem.status]} with {len(pool)} people available "
            f"for {size} places"
        )

    members = sorted(pool[i] for i in range(len(pool)) if y[i].value() > 0.5)
    assignments = {
        events[j]: sorted(
            pool[i] for i in range(len(pool)) if x[i, j].value() > 0.5
        )
        for j in range(len(events))
    }

    return members, assignments, float(pulp.value(problem.objective))


def build_teams(
    grade_levels: Mapping[str, int],
    preferences: Mapping[str, Sequence[str]],
    scores: Mapping[str, Mapping[str, float]],
    years_experience: Mapping[str, float],
    event_time_blocks: Mapping[str, Any],
    events: Iterable[str],
    event_capacity: Mapping[str, int],
    team_size: int = 15,
    max_events_per_person: int = 4,
    max_seniors_per_team: int = 7,
    everyone_competes: bool = True,
    max_experience_by_grade: Mapping[int, int] = MAX_EXPERIENCE_BY_GRADE,
    solver: Any = None,
) -> Roster:
    """Place everyone on team A, B or C and staff every event to capacity.

    Args:
        grade_levels: person -> 9, 10, 11 or 12. Used only for the senior cap,
            never for ranking.
        preferences: person -> their ranked event wish list (five, by convention);
            earlier entries count for more.
        scores: person -> (event -> score) from other Science Olympiad events.
            Scores are normalised per event, so differing scales are comparable.
        years_experience: person -> seasons of Science Olympiad behind them.
            Independent of grade level, and weighed when choosing team members
            rather than when choosing their events.
        event_time_blocks: event -> time block. Two events clash exactly when
            their blocks are equal.
        events: the events to staff on the day.
        event_capacity: event -> how many people **each team** enters.
        team_size: members on A and B (C absorbs everyone left over).
        max_events_per_person: cap on events per person.
        max_seniors_per_team: cap on 12th graders for A and B. Team C is the
            overflow team and is exempt, so it may hold any number of seniors.
        everyone_competes: require every team member to enter at least one event.
            On by default; without it one strong competitor covers several events
            and the remaining places are padded with people who never compete.
        max_experience_by_grade: plausible ceiling on seasons for each grade,
            used to reject impossible data.
        solver: a pulp solver; defaults to bundled CBC, quiet.

    Returns:
        A :class:`Roster` with the split, the per-team event assignments, and the
        objective value each team achieved.

    Raises:
        ValueError: if the inputs are malformed, or the constraints cannot all be
            met (the message carries the arithmetic that rules it out).
    """
    events = list(events)
    people = _people(grade_levels, preferences, scores, years_experience)
    solver = solver or default_solver()

    sizes = {
        "A": team_size,
        "B": team_size,
        "C": len(people) - 2 * team_size,
    }
    senior_caps = _senior_caps(sizes, max_seniors_per_team)
    _validate(
        people,
        events,
        grade_levels,
        years_experience,
        max_experience_by_grade,
        event_time_blocks,
        event_capacity,
        sizes,
        senior_caps,
        max_events_per_person,
        everyone_competes,
    )

    blocks: dict[Any, list[str]] = {}
    for event in events:
        blocks.setdefault(event_time_blocks[event], []).append(event)

    score_index = normalised_scores(people, events, scores)
    pref_index = preference_index(people, preferences)
    experience = normalised_experience(people, years_experience)

    roster = Roster(score_index=score_index, experience_index=experience)
    remaining = list(people)
    for position, team in enumerate(TEAMS):
        members, assignments, objective = _solve_team(
            team,
            remaining,
            sizes[team],
            senior_caps[team],
            [(sizes[nxt], senior_caps[nxt]) for nxt in TEAMS[position + 1 :]],
            events,
            event_capacity,
            blocks,
            score_index,
            pref_index,
            experience,
            grade_levels,
            max_events_per_person,
            everyone_competes,
            solver,
        )
        roster.teams[team] = members
        roster.event_assignments[team] = assignments
        roster.objective[team] = objective
        for person in members:
            roster.team_of[person] = team
        remaining = [p for p in remaining if p not in set(members)]

    roster.person_events = {
        person: sorted(
            event
            for event, competitors in roster.event_assignments[
                roster.team_of[person]
            ].items()
            if person in competitors
        )
        for person in people
    }

    roster.reserves = sorted(
        person for person, entered in roster.person_events.items() if not entered
    )

    _verify(
        roster,
        people,
        events,
        grade_levels,
        event_time_blocks,
        event_capacity,
        sizes,
        senior_caps,
        max_events_per_person,
        everyone_competes,
    )
    return roster


def _verify(
    roster: Roster,
    people: Sequence[str],
    events: Sequence[str],
    grade_levels: Mapping[str, int],
    event_time_blocks: Mapping[str, Any],
    event_capacity: Mapping[str, int],
    sizes: Mapping[str, int],
    senior_caps: Mapping[str, int],
    max_events_per_person: int,
    everyone_competes: bool,
) -> None:
    """Re-check every criterion against the solver's answer, independently.

    The ILP should make all of this true by construction; checking anyway is what
    turns a modelling mistake into a loud failure instead of a plausible roster.
    """
    placed = [p for team in TEAMS for p in roster.teams[team]]
    if sorted(placed) != sorted(people):
        raise ValueError("every person must be on exactly one team")
    if len(placed) != len(set(placed)):
        raise ValueError("a person appears on more than one team")

    for team in TEAMS:
        if len(roster.teams[team]) != sizes[team]:
            raise ValueError(
                f"team {team} has {len(roster.teams[team])} members, expected {sizes[team]}"
            )
        seniors = sum(
            1 for p in roster.teams[team] if grade_levels[p] == SENIOR_GRADE
        )
        if seniors > senior_caps[team]:
            raise ValueError(
                f"team {team} has {seniors} twelfth graders (max {senior_caps[team]})"
            )

        for event in events:
            competitors = roster.event_assignments[team][event]
            if len(competitors) != event_capacity[event]:
                raise ValueError(
                    f"team {team} put {len(competitors)} people in {event!r}, "
                    f"expected exactly {event_capacity[event]}"
                )
            if len(set(competitors)) != len(competitors):
                raise ValueError(f"duplicate competitor in {team}/{event!r}")
            outsiders = set(competitors) - set(roster.teams[team])
            if outsiders:
                raise ValueError(
                    f"{sorted(outsiders)} compete for team {team} without being on it"
                )

    for person, person_events in roster.person_events.items():
        if len(person_events) > max_events_per_person:
            raise ValueError(
                f"{person} is in {len(person_events)} events (max {max_events_per_person})"
            )
        team = roster.team_of[person]
        fits = sizes[team] <= sum(event_capacity[e] for e in events)
        if everyone_competes and fits and not person_events:
            raise ValueError(
                f"{person} is on team {team}, which has room for everyone to "
                "compete, but competes in nothing"
            )

        seen_blocks: dict[Any, str] = {}
        for event in person_events:
            block = event_time_blocks[event]
            if block in seen_blocks:
                raise ValueError(
                    f"{person} is double-booked in time block {block!r}: "
                    f"{seen_blocks[block]!r} and {event!r}"
                )
            seen_blocks[block] = event
````

````python
"""Tests for :mod:`so_team_builder`.

Run with ``pytest test_so_team_builder.py`` or ``python test_so_team_builder.py``.
"""

from __future__ import annotations

import random

import pytest
'''
from so_team_builder import (
    MAX_EXPERIENCE_BY_GRADE,
    SENIOR_GRADE,
    TEAM_WEIGHTS,
    TEAMS,
    Roster,
    build_score_buckets,
    build_teams,
    normalised_experience,
    normalised_scores,
    preference_index,
)
'''
# Ten events over five time blocks: the two events sharing a block clash, so
# nobody can hold both.
EVENT_TIME_BLOCKS = EventTimeSlots.DIV_C_BLOCKS
EVENTS = current_events

# Each team enters two people per event: 10 events x 2 = 20 places per team,
# comfortably inside 15 members x 3 events = 45.
''' EVENT_CAPACITY = {event: 2 for event in EVENTS} '''

TEAM_SIZE = 15
PEOPLE_COUNT = len(all_people)

def make_scenario(seed: int = 7):
    """A full 45-person school: three teams of fifteen.

    Exactly 21 twelfth graders, which is the most the 7-per-team cap allows, so
    the senior constraint is tight rather than decorative.
    """
    rng = random.Random(seed)
    people = [f"student{i:02d}" for i in range(PEOPLE_COUNT)]

    grade_levels = {}
    for index, person in enumerate(people):
        grade_levels[person] = SENIOR_GRADE if index < 21 else rng.choice([9, 10, 11])

    # Experience is drawn independently of grade level — bounded only by what is
    # possible for that grade — so the squad contains both 12th-grade first-timers
    # and 9th graders who have already served their maximum two seasons.
    years_experience = {
        person: rng.randint(0, MAX_EXPERIENCE_BY_GRADE[grade_levels[person]])
        for person in people
    }

    # Everyone has competed in some events but nobody in all of them.
    scores = {
        person: {
            event: round(rng.uniform(40, 100), 1)
            for event in EVENTS
            if rng.random() < 0.55
        }
        for person in people
    }

    preferences = {person: rng.sample(EVENTS, k=5) for person in people}
    return grade_levels, preferences, scores, years_experience

def make_true_scenario():
    grade_levels = {
        person.name: person.grade for person in all_people
    }

    preferences = {
        person.name: person.chosen_events for person in all_people
    }

    scores = {
        person.name: person.scores for person in all_people
    }

    years_experience = {
        person.name: person.years_of_participation for person in all_people
    }
    return grade_levels, preferences, scores, years_experience

GRADE_LEVELS, PREFERENCES, SCORES, EXPERIENCE = make_true_scenario()

def build(**kwargs) -> Roster:
    return build_teams(
        GRADE_LEVELS, PREFERENCES, SCORES, EXPERIENCE, EVENT_TIME_BLOCKS, EVENTS,
        EVENT_CAPACITY, team_size=TEAM_SIZE, **kwargs
    )

@pytest.fixture(scope="module")
def roster() -> Roster:
    return build()

# ---------------------------------------------------------------------------
# Output criteria 1-7
# ---------------------------------------------------------------------------

def test_everyone_is_on_exactly_one_team(roster):
    placed = [p for team in TEAMS for p in roster.teams[team]]
    assert sorted(placed) == sorted(GRADE_LEVELS)
    assert len(placed) == len(set(placed))

def test_a_and_b_hold_fifteen_and_c_absorbs_the_rest(roster):
    assert len(roster.teams["A"]) == TEAM_SIZE
    assert len(roster.teams["B"]) == TEAM_SIZE
    assert len(roster.teams["C"]) == PEOPLE_COUNT - 2 * TEAM_SIZE

def test_a_and_b_carry_at_most_seven_twelfth_graders(roster):
    """C is the overflow team and is exempt, so the cap binds only on A and B."""
    for team in ("A", "B"):
        seniors = [p for p in roster.teams[team] if GRADE_LEVELS[p] == SENIOR_GRADE]
        assert len(seniors) <= 7, f"team {team} has {len(seniors)} seniors"

def test_nobody_holds_two_events_in_one_time_block(roster):
    for person, events in roster.person_events.items():
        blocks = [EVENT_TIME_BLOCKS[event] for event in events]
        assert len(blocks) == len(set(blocks)), f"{person} clashes: {events}"

def test_nobody_competes_in_more_than_three_events(roster):
    for person, events in roster.person_events.items():
        assert len(events) <= 3, f"{person} has {len(events)} events"

def test_every_event_is_staffed_by_every_team(roster):
    for team in TEAMS:
        for event in EVENTS:
            assert roster.event_assignments[team][event], f"{team}/{event} empty"

def test_every_event_is_filled_to_exactly_its_capacity(roster):
    for team in TEAMS:
        for event in EVENTS:
            competitors = roster.event_assignments[team][event]
            assert len(competitors) == EVENT_CAPACITY[event]

def test_competitors_belong_to_the_team_they_compete_for(roster):
    for team in TEAMS:
        members = set(roster.teams[team])
        for event, competitors in roster.event_assignments[team].items():
            assert set(competitors) <= members

def test_every_member_competes_in_at_least_one_event(roster):
    """No warm bodies: a place on the team means competing.

    Without this the model is degenerate -- one strong person covers several
    events and the leftover places cost nothing, so they get padded with whoever.
    """
    for person, events in roster.person_events.items():
        assert events, f"{person} is on team {roster.team_of[person]} but sits out"

def test_relaxing_everyone_competes_allows_one_person_to_cover_several():
    """The opt-out is what the constraint is protecting against, made explicit.

    ace outscores everyone, so with the requirement lifted the optimiser enters
    him in both events and the second place on team A goes to a spectator.
    """
    events = ["Optics", "Ecology"]
    blocks = {"Optics": 0, "Ecology": 1}
    capacity = {"Optics": 1, "Ecology": 1}
    names = ["ace", "bolt", "cat", "dot", "elm", "fig"]
    scores = {
        name: {"Optics": 99 - i * 20, "Ecology": 98 - i * 20}
        for i, name in enumerate(names)
    }
    roster = build_teams(
        dict.fromkeys(names, 10), {}, scores, dict.fromkeys(names, 2),
        blocks, events, capacity, team_size=2, everyone_competes=False
    )
    assert roster.person_events["ace"] == ["Ecology", "Optics"]
    assert any(not roster.person_events[p] for p in roster.teams["A"])

def test_rejects_fixed_teams_larger_than_the_places_available():
    """A and B are pinned at 15, so four places for them is a configuration error."""
    events = ["a", "b"]
    blocks = {"a": 0, "b": 1}
    capacity = {"a": 2, "b": 2}
    grade_levels = {f"p{i}": 10 for i in range(45)}
    with pytest.raises(ValueError, match="would sit out"):
        build_teams(
            grade_levels, {}, {}, dict.fromkeys(grade_levels, 1),
            blocks, events, capacity, team_size=15,
        )

def _big_school(people_count: int):
    """A squad too large for every member to compete, to exercise the overflow."""
    people = [f"p{i:02d}" for i in range(people_count)]
    scores = {
        p: {e: (i * 7 + j * 13) % 100 for j, e in enumerate(EVENTS)}
        for i, p in enumerate(people)
    }
    return (
        dict.fromkeys(people, 10),
        {p: [] for p in people},
        scores,
        dict.fromkeys(people, 1),
    )

def test_overflow_team_c_may_hold_far_more_than_fifteen():
    """60 students: A and B take 15 each and C absorbs the remaining 30."""
    grade_levels, preferences, scores, experience = _big_school(60)
    roster = build_teams(
        grade_levels, preferences, scores, experience, EVENT_TIME_BLOCKS, EVENTS,
        EVENT_CAPACITY, team_size=TEAM_SIZE,
    )
    assert len(roster.teams["A"]) == len(roster.teams["B"]) == 15
    assert len(roster.teams["C"]) == 30
    assert sorted(p for team in TEAMS for p in roster.teams[team]) == sorted(grade_levels)

def test_surplus_members_are_reported_as_reserves_on_team_c_only():
    """Only 20 places per team, so 10 of C's 30 must sit out -- and are named."""
    grade_levels, preferences, scores, experience = _big_school(60)
    roster = build_teams(
        grade_levels, preferences, scores, experience, EVENT_TIME_BLOCKS, EVENTS,
        EVENT_CAPACITY, team_size=TEAM_SIZE,
    )
    places = sum(EVENT_CAPACITY.values())
    assert len(roster.reserves) == 30 - places
    assert all(roster.team_of[p] == "C" for p in roster.reserves)
    # A and B are inside the places available, so all of them compete.
    for team in ("A", "B"):
        assert all(roster.person_events[p] for p in roster.teams[team])

def test_an_oversubscribed_team_spreads_the_work_as_widely_as_it_can():
    """With more members than places, nobody doubles up while others sit idle."""
    grade_levels, preferences, scores, experience = _big_school(60)
    roster = build_teams(
        grade_levels, preferences, scores, experience, EVENT_TIME_BLOCKS, EVENTS,
        EVENT_CAPACITY, team_size=TEAM_SIZE,
    )
    competing = [p for p in roster.teams["C"] if roster.person_events[p]]
    assert len(competing) == sum(EVENT_CAPACITY.values())
    assert all(len(roster.person_events[p]) == 1 for p in competing)


# ---------------------------------------------------------------------------
# Ranking criteria 8-9
# ---------------------------------------------------------------------------

def _mean_score_of_assignments(roster, team):
    """Average normalised score of the people actually fielded by a team."""
    pairs = [
        roster.score_index[person].get(event, 0.0)
        for event, competitors in roster.event_assignments[team].items()
        for person in competitors
    ]
    return sum(pairs) / len(pairs)

def test_team_a_fields_the_strongest_scorers_then_b_then_c(roster):
    """Criterion 8: A takes the best pairings, B the next, C the remainder."""
    a = _mean_score_of_assignments(roster, "A")
    b = _mean_score_of_assignments(roster, "B")
    c = _mean_score_of_assignments(roster, "C")
    assert a > b > c, f"expected A > B > C, got A={a:.3f} B={b:.3f} C={c:.3f}"

def test_team_c_has_the_most_preferences_granted(roster):
    """Criterion 9: preferences dominate on C, and count least on A."""
    granted = {}
    for team in TEAMS:
        granted[team] = sum(
            1
            for event, competitors in roster.event_assignments[team].items()
            for person in competitors
            if event in PREFERENCES[person]
        )
    per_person = {team: granted[team] / len(roster.teams[team]) for team in TEAMS}
    assert per_person["C"] > per_person["A"], (
        f"C should honour more preferences per head than A: {per_person}"
    )
    assert TEAM_WEIGHTS["C"][1] > TEAM_WEIGHTS["B"][1] > TEAM_WEIGHTS["A"][1]

def test_the_solution_is_optimal_not_merely_valid(roster):
    """Each team's objective is the best achievable, so no swap can improve it."""
    assert roster.objective["A"] > 0
    # A is optimised on score with only a light preference term, so its objective
    # must beat what B achieves on the leftovers.
    assert roster.objective["A"] > roster.objective["B"]

def test_scores_are_normalised_per_event():
    """Two events on different scales end up equally weighted."""
    people = ["p1", "p2"]
    events = ["small", "large"]
    index = normalised_scores(
        people, events, {"p1": {"small": 5, "large": 100}, "p2": {"small": 1, "large": 20}}
    )
    assert index["p1"]["small"] == index["p1"]["large"] == 1.0
    assert index["p2"]["small"] == index["p2"]["large"] == pytest.approx(0.1)

def test_missing_score_ranks_below_any_recorded_score():
    """A recorded score always beats no record, even a low one."""
    index = normalised_scores(["p1", "p2", "p3"], ["e"], {"p1": {"e": 90}, "p2": {"e": 1}})
    assert index["p1"]["e"] == 1.0
    assert index["p2"]["e"] == pytest.approx(0.1)  # worst recorded, still above zero
    assert index["p3"].get("e", 0.0) == 0.0        # no record at all

def test_a_lone_recorded_score_normalises_to_the_top():
    """With nothing to compare against, the only scorer is the best scorer."""
    index = normalised_scores(["p1", "p2"], ["e"], {"p1": {"e": 1}})
    assert index["p1"]["e"] == 1.0
    assert index["p2"].get("e", 0.0) == 0.0

def test_all_five_preferences_count_equally():
    """The wish list is a set of five, not a ranking, so order carries no meaning."""
    index = preference_index(["p"], {"p": ["first", "second", "third", "fourth", "fifth"]})
    assert set(index["p"].values()) == {1.0}
    assert index["p"]["first"] == index["p"]["fifth"] == 1.0
    assert "unlisted" not in index["p"]

def test_reordering_a_wish_list_changes_nothing():
    """Two people asking for the same five events are indistinguishable."""
    forwards = preference_index(["p"], {"p": ["a", "b", "c", "d", "e"]})
    backwards = preference_index(["p"], {"p": ["e", "d", "c", "b", "a"]})
    assert forwards == backwards


# ---------------------------------------------------------------------------
# A small, hand-checkable instance
# ---------------------------------------------------------------------------

def test_tiny_instance_puts_the_top_scorers_on_a():
    """Six people, two events in separate blocks, one place each per team.

    ace and bolt are far and away the strongest, so they must form team A.
    """
    events = ["Optics", "Ecology"]
    blocks = {"Optics": 0, "Ecology": 1}
    capacity = {"Optics": 1, "Ecology": 1}
    grade_levels = dict.fromkeys(["ace", "bolt", "cat", "dot", "elm", "fig"], 10)
    scores = {
        "ace": {"Optics": 99, "Ecology": 98},
        "bolt": {"Optics": 97, "Ecology": 96},
        "cat": {"Optics": 70, "Ecology": 71},
        "dot": {"Optics": 68, "Ecology": 69},
        "elm": {"Optics": 30, "Ecology": 31},
        "fig": {"Optics": 28, "Ecology": 29},
    }
    roster = build_teams(
        grade_levels, {}, scores, dict.fromkeys(grade_levels, 2),
        blocks, events, capacity, team_size=2
    )
    assert roster.teams["A"] == ["ace", "bolt"]
    assert roster.teams["B"] == ["cat", "dot"]
    assert roster.teams["C"] == ["elm", "fig"]

def test_preferences_break_ties_on_team_c():
    """With scores equal, C is placed on what people asked for."""
    events = ["Optics", "Ecology"]
    blocks = {"Optics": 0, "Ecology": 1}
    capacity = {"Optics": 1, "Ecology": 1}
    names = ["ace", "bolt", "cat", "dot", "elm", "fig"]
    grade_levels = dict.fromkeys(names, 10)
    # Identical scores and experience, so only preferences can decide C's line-up.
    scores = {name: {"Optics": 50, "Ecology": 50} for name in names}
    preferences = {"elm": ["Ecology"], "fig": ["Optics"]}
    roster = build_teams(
        grade_levels, preferences, scores, dict.fromkeys(names, 2),
        blocks, events, capacity, team_size=2
    )
    c_assignments = roster.event_assignments["C"]
    if set(roster.teams["C"]) == {"elm", "fig"}:
        assert c_assignments["Ecology"] == ["elm"]
        assert c_assignments["Optics"] == ["fig"]


# ---------------------------------------------------------------------------
# Years of experience
# ---------------------------------------------------------------------------

TWO_EVENTS = ["Optics", "Ecology"]
TWO_BLOCKS = {"Optics": 0, "Ecology": 1}
ONE_EACH = {"Optics": 1, "Ecology": 1}

def test_experience_decides_between_people_with_identical_scores():
    """Scores tied everywhere, so the veterans are the ones promoted to A."""
    names = ["ace", "bolt", "cat", "dot", "elm", "fig"]
    scores = {name: {"Optics": 50, "Ecology": 50} for name in names}
    experience = {"ace": 3, "bolt": 3, "cat": 2, "dot": 2, "elm": 0, "fig": 0}
    roster = build_teams(
        dict.fromkeys(names, 10), {}, scores, experience,
        TWO_BLOCKS, TWO_EVENTS, ONE_EACH, team_size=2
    )
    assert roster.teams["A"] == ["ace", "bolt"]
    assert roster.teams["C"] == ["elm", "fig"]

def test_experience_does_not_override_a_clear_score_gap():
    """A rookie who is far stronger still makes team A over a veteran."""
    names = ["rookie", "veteran", "cat", "dot", "elm", "fig"]
    scores = {
        "rookie": {"Optics": 100, "Ecology": 100},
        "veteran": {"Optics": 20, "Ecology": 20},
        **{n: {"Optics": 50, "Ecology": 50} for n in ["cat", "dot", "elm", "fig"]},
    }
    experience = {"rookie": 0, "veteran": 3, "cat": 0, "dot": 0, "elm": 0, "fig": 0}
    roster = build_teams(
        dict.fromkeys(names, 10), {}, scores, experience,
        TWO_BLOCKS, TWO_EVENTS, ONE_EACH, team_size=2
    )
    assert "rookie" in roster.teams["A"]
    assert "veteran" not in roster.teams["A"]

def test_twelfth_grade_rookie_ranks_below_a_ninth_grader_with_two_seasons():
    """Grade level is not experience: seasons served are what count.

    Two is all a 9th grader can have, and it still beats a 12th grader's none.
    """
    names = ["senior_rookie", "junior_veteran", "cat", "dot", "elm", "fig"]
    grade_levels = {
        "senior_rookie": 12, "junior_veteran": 9,
        "cat": 10, "dot": 10, "elm": 10, "fig": 10,
    }
    scores = {name: {"Optics": 50, "Ecology": 50} for name in names}
    experience = {
        "senior_rookie": 0, "junior_veteran": 2,
        "cat": 1, "dot": 1, "elm": 1, "fig": 1,
    }
    roster = build_teams(
        grade_levels, {}, scores, experience,
        TWO_BLOCKS, TWO_EVENTS, ONE_EACH, team_size=2
    )
    assert "junior_veteran" in roster.teams["A"]
    assert "senior_rookie" not in roster.teams["A"]

def test_rejects_experience_impossible_for_the_grade_level():
    """A 9th grader with four seasons is a typo, not a prodigy."""
    names = ["ace", "bolt", "cat", "dot", "elm", "fig"]
    grade_levels = dict.fromkeys(names, 10) | {"ace": 9}
    scores = {name: {"Optics": 50, "Ecology": 50} for name in names}
    experience = dict.fromkeys(names, 1) | {"ace": 4}
    with pytest.raises(ValueError, match="impossible for grade level"):
        build_teams(
            grade_levels, {}, scores, experience,
            TWO_BLOCKS, TWO_EVENTS, ONE_EACH, team_size=2
        )

def test_grade_level_ceilings_are_what_the_domain_expects():
    """A 9th grader tops out at two seasons, each later grade adds one."""
    assert MAX_EXPERIENCE_BY_GRADE == {9: 2, 10: 3, 11: 4, 12: 5}

def test_uniform_experience_has_no_effect():
    """When everyone has served the same time the term cannot sway the choice."""
    names = ["ace", "bolt", "cat", "dot", "elm", "fig"]
    grade_levels = dict.fromkeys(names, 10)
    scores = {
        name: {"Optics": 90 - i * 10, "Ecology": 90 - i * 10}
        for i, name in enumerate(names)
    }
    green = build_teams(grade_levels, {}, scores, dict.fromkeys(names, 0),
                        TWO_BLOCKS, TWO_EVENTS, ONE_EACH, team_size=2)
    grizzled = build_teams(grade_levels, {}, scores, dict.fromkeys(names, 3),
                           TWO_BLOCKS, TWO_EVENTS, ONE_EACH, team_size=2)
    assert green.teams == grizzled.teams

def test_experience_is_normalised_across_the_squad():
    """Experience is relative: the most seasoned person scores 1.0, the least 0.0."""
    # from so_team_builder import normalised_experience
    index = normalised_experience(["a", "b", "c"], {"a": 0, "b": 2, "c": 4})
    assert index == {"a": 0.0, "b": 0.5, "c": 1.0}

def test_teams_trend_towards_more_experience_at_the_top(roster):
    """Across the full 45-person school, A is the most seasoned team."""
    mean = {
        team: sum(EXPERIENCE[p] for p in roster.teams[team]) / len(roster.teams[team])
        for team in TEAMS
    }
    assert mean["A"] > mean["C"], f"A should be more experienced than C: {mean}"


# ---------------------------------------------------------------------------
# Score buckets from medals and coursework
# ---------------------------------------------------------------------------

COURSE_EVENTS = {
    # AP Chemistry prepares fully for Chemistry Lab, partly for the others.
    "AP Chemistry": {"Chemistry Lab": 1.0, "Materials Science": 0.7, "Forensics": 0.5},
    "AP Biology": {"Anatomy & Physiology", "Disease Detectives"}, # unweighted = 1.0
}

def test_medals_and_coursework_both_feed_the_bucket():
    buckets = build_score_buckets(
        medals={"ace": [{"event": "Chemistry Lab", "place": 1}]},
        coursework={"ace": ["AP Chemistry"]},
        course_events=COURSE_EVENTS,
        course_points=2.0,
    )
    # The gold stands, with study adding a little on top but never a whole medal.
    assert 10.0 < buckets["ace"]["Chemistry Lab"] < 11.0
    # The same course spills into related events at the strength given for each.
    assert buckets["ace"]["Materials Science"] > buckets["ace"]["Forensics"] > 0

def test_any_medal_outranks_any_amount_of_study():
    """A single fifth place beats a pile of relevant classes."""
    buckets = build_score_buckets(
        medals={"competitor": [{"event": "Chemistry Lab", "place": 5}]},
        coursework={"student": ["AP Chemistry"] * 20},
        course_events=COURSE_EVENTS,
    )
    assert buckets["competitor"]["Chemistry Lab"] > buckets["student"]["Chemistry Lab"]

def test_study_cannot_reorder_two_medallists():
    """A bronze plus every class going still sits below a silver."""
    buckets = build_score_buckets(
        medals={
            "silver": [{"event": "Chemistry Lab", "place": 2}],
            "bronze": [{"event": "Chemistry Lab", "place": 3}],
        },
        coursework={"bronze": ["AP Chemistry"] * 20},
        course_events=COURSE_EVENTS,
    )
    assert buckets["silver"]["Chemistry Lab"] > buckets["bronze"]["Chemistry Lab"]

def test_study_still_separates_people_who_never_medalled():
    """Among non-medallists, coursework is the only evidence there is."""
    buckets = build_score_buckets(
        medals={},
        coursework={"keen": ["AP Chemistry", "AP Biology"], "less_keen": ["AP Biology"]},
        course_events=COURSE_EVENTS,
    )
    keen = buckets["keen"]["Anatomy & Physiology"]
    less = buckets["less_keen"]["Anatomy & Physiology"]
    assert keen == less # same single course feeds this event for both
    assert buckets["keen"]["Chemistry Lab"] > 0
    assert "Chemistry Lab" not in buckets["less_keen"]

def test_study_breaks_ties_between_equal_medallists():
    """Two bronzes, and the one who also took the class edges ahead."""
    buckets = build_score_buckets(
        medals={
            "studious": [{"event": "Chemistry Lab", "place": 3}],
            "idle": [{"event": "Chemistry Lab", "place": 3}],
        },
        coursework={"studious": ["AP Chemistry"]},
        course_events=COURSE_EVENTS,
    )
    assert buckets["studious"]["Chemistry Lab"] > buckets["idle"]["Chemistry Lab"]

def test_additive_mode_lets_coursework_outweigh_a_medal():
    """The opt-out, shown explicitly: this is what the default is guarding against."""
    buckets = build_score_buckets(
        medals={"competitor": [{"event": "Chemistry Lab", "place": 1}]},
        coursework={"student": ["AP Chemistry"] * 20},
        course_events=COURSE_EVENTS,
        course_points=2.0,
        medals_outrank_coursework=False,
    )
    assert buckets["student"]["Chemistry Lab"] > buckets["competitor"]["Chemistry Lab"]

def test_repeat_medals_stack():
    buckets = build_score_buckets(
        medals={
            "ace": [
                {"event": "Optics", "place": 1},
                {"event": "Optics", "place": 3},
            ],
        },
        coursework={},
        course_events={},
    )
    assert buckets["ace"]["Optics"] == 15.0  # 10 + 5

def test_unplaced_results_contribute_nothing():
    """Only placings in the medal table score; a 20th place is not evidence."""
    buckets = build_score_buckets(
        medals={"ace": [{"event": "Optics", "place": 20}]},
        coursework={},
        course_events={},
    )
    assert not buckets["ace"]

def test_unweighted_course_feeds_every_listed_event_equally():
    """A course listed without weights counts fully towards each event it names."""
    buckets = build_score_buckets(
        medals={}, coursework={"ace": ["AP Biology"]},
        course_events=COURSE_EVENTS, course_points=3.0,
    )
    assert set(buckets["ace"]) == {"Anatomy & Physiology", "Disease Detectives"}
    assert len(set(buckets["ace"].values())) == 1  # both events weighted alike

def test_relevance_map_is_the_dial_that_still_bites():
    """course_points normalises away, so relevance decides how study is spread."""
    args = dict(medals={}, coursework={"ace": ["AP Chemistry"]},
                course_events=COURSE_EVENTS)
    light = build_score_buckets(**args, course_points=1.0)
    heavy = build_score_buckets(**args, course_points=20.0)
    assert light["ace"] == pytest.approx(heavy["ace"])
    # But the per-event relevance still orders the spill-over.
    chem = light["ace"]["Chemistry Lab"]
    assert chem > light["ace"]["Materials Science"] > light["ace"]["Forensics"]

def test_course_points_is_a_real_dial_in_additive_mode():
    """With the tier lifted, course_points trades directly against the medal table."""
    args = dict(
        medals={"ace": [{"event": "Chemistry Lab", "place": 3}]},
        coursework={"ace": ["AP Chemistry"]},
        course_events=COURSE_EVENTS,
        medals_outrank_coursework=False,
    )
    light = build_score_buckets(**args, course_points=1.0)
    heavy = build_score_buckets(**args, course_points=20.0)
    assert light["ace"]["Chemistry Lab"] == 6.0  # bronze 5 + one course
    assert heavy["ace"]["Chemistry Lab"] == 25.0  # coursework dominates

def test_unknown_course_is_rejected_rather_than_ignored():
    """A course with no mapping would otherwise silently score nothing."""
    with pytest.raises(ValueError, match="not in course_events"):
        build_score_buckets(
            medals={}, coursework={"ace": ["Underwater Basket Weaving"]},
            course_events=COURSE_EVENTS,
        )

def test_buckets_feed_straight_into_build_teams():
    """The builder's output is exactly the 'scores' shape build_teams consumes."""
    names = ["ace", "bolt", "cat", "dot", "elm", "fig"]
    buckets = build_score_buckets(
        medals={
            name: [
                {"event": "Optics", "place": place},
                {"event": "Ecology", "place": place},
            ]
            for name, place in zip(names, [1, 2, 3, 4, 5, 20])
        },
        coursework={name: [] for name in names},
        course_events=COURSE_EVENTS,
    )
    roster = build_teams(
        dict.fromkeys(names, 10), {}, buckets, dict.fromkeys(names, 1),
        TWO_BLOCKS, TWO_EVENTS, ONE_EACH, team_size=2,
    )
    assert roster.teams["A"] == ["ace", "bolt"]  # gold and silver


# ---------------------------------------------------------------------------
# Infeasible and malformed input
# ---------------------------------------------------------------------------

def _with_seniors(count: int) -> tuple[dict[str, int], dict[str, int]]:
    """The standard scenario re-graded so exactly 'count' people are seniors.

    Experience is clamped to whatever the new grade allows, since re-grading a
    12th grader down to 10th would otherwise leave them with impossible seasons.
    """
    grade_levels = {
        person: SENIOR_GRADE if index < count else 10
        for index, person in enumerate(sorted(GRADE_LEVELS))
    }
    experience = {
        person: min(EXPERIENCE[person], MAX_EXPERIENCE_BY_GRADE[grade])
        for person, grade in grade_levels.items()
    }
    return grade_levels, experience

def test_overflow_team_c_may_hold_more_than_seven_twelfth_graders():
    """25 seniors: A and B take 7 each at most, C absorbs the remaining 11+."""
    grade_levels, experience = _with_seniors(25)
    roster = build_teams(
        grade_levels, PREFERENCES, SCORES, experience, EVENT_TIME_BLOCKS, EVENTS,
        EVENT_CAPACITY, team_size=TEAM_SIZE,
    )
    counts = {
        team: sum(1 for p in roster.teams[team] if grade_levels[p] == SENIOR_GRADE)
        for team in TEAMS
    }
    assert counts["A"] <= 7 and counts["B"] <= 7
    assert counts["C"] > 7, f"C should absorb the overflow, got {counts}"
    assert sum(counts.values()) == 25

def test_seniors_beyond_what_even_the_overflow_team_can_hold_are_rejected():
    """A(7) + B(7) + C(15 seats) = 29 places for seniors; 30 cannot be placed."""
    grade_levels, experience = _with_seniors(30)
    with pytest.raises(ValueError, match="twelfth graders cannot be placed"):
        build_teams(
            grade_levels, PREFERENCES, SCORES, experience, EVENT_TIME_BLOCKS, EVENTS,
            EVENT_CAPACITY, team_size=TEAM_SIZE,
        )

def test_all_seniors_on_the_overflow_team_is_allowed():
    """29 seniors is exactly the ceiling: 7 + 7 + a full team C."""
    grade_levels, experience = _with_seniors(29)
    roster = build_teams(
        grade_levels, PREFERENCES, SCORES, experience, EVENT_TIME_BLOCKS, EVENTS,
        EVENT_CAPACITY, team_size=TEAM_SIZE,
    )
    assert all(grade_levels[p] == SENIOR_GRADE for p in roster.teams["C"])

def test_rejects_a_time_block_bigger_than_a_team():
    """Three events in one block needing 2 each cannot be covered by 5 people."""
    events = ["a", "b", "c"]
    blocks = dict.fromkeys(events, 0)  # all clash with each other
    capacity = dict.fromkeys(events, 2)
    grade_levels = {f"p{i}": 10 for i in range(15)}
    experience = dict.fromkeys(grade_levels, 1)
    with pytest.raises(ValueError, match="needs 6 people at once"):
        build_teams(grade_levels, {}, {}, experience, blocks, events, capacity, team_size=5)

def test_rejects_more_places_than_the_team_can_supply():
    """Ten events in ten blocks, 2 each, is 20 places -- beyond 3 people x 3 events."""
    events = [f"e{i}" for i in range(10)]
    blocks = {event: index for index, event in enumerate(events)}
    capacity = dict.fromkeys(events, 2)
    grade_levels = {f"p{i}": 10 for i in range(9)}
    experience = dict.fromkeys(grade_levels, 1)
    with pytest.raises(ValueError, match="can supply only"):
        build_teams(grade_levels, {}, {}, experience, blocks, events, capacity, team_size=3)

def test_rejects_zero_capacity_event():
    """Criterion 6: every event must get somebody."""
    with pytest.raises(ValueError, match="at least 1"):
        build_teams(
            GRADE_LEVELS, PREFERENCES, SCORES, EXPERIENCE, EVENT_TIME_BLOCKS, EVENTS,
            {**EVENT_CAPACITY, "Optics": 0}, team_size=TEAM_SIZE,
        )

def test_rejects_event_with_no_time_block():
    with pytest.raises(ValueError, match="no time block"):
        build_teams(
            GRADE_LEVELS, PREFERENCES, SCORES, EXPERIENCE,
            {k: v for k, v in EVENT_TIME_BLOCKS.items() if k != "Optics"},
            EVENTS, EVENT_CAPACITY, team_size=TEAM_SIZE,
        )

def test_rejects_missing_or_invalid_years_of_experience():
    """A mistyped name must not silently become a first-year."""
    experience = dict(EXPERIENCE)
    del experience["student00"]
    with pytest.raises(ValueError, match="years_experience"):
        build_teams(
            GRADE_LEVELS, PREFERENCES, SCORES, experience, EVENT_TIME_BLOCKS,
            EVENTS, EVENT_CAPACITY, team_size=TEAM_SIZE,
        )

    negative = {**EXPERIENCE, "student00": -1}
    with pytest.raises(ValueError, match="years_experience"):
        build_teams(
            GRADE_LEVELS, PREFERENCES, SCORES, negative, EVENT_TIME_BLOCKS,
            EVENTS, EVENT_CAPACITY, team_size=TEAM_SIZE,
        )

def test_rejects_missing_or_invalid_grade_level():
    grade_levels = dict(GRADE_LEVELS)
    grade_levels["student00"] = 13
    with pytest.raises(ValueError, match="grade_level"):
        build_teams(
            grade_levels, PREFERENCES, SCORES, EXPERIENCE, EVENT_TIME_BLOCKS, EVENTS,
            EVENT_CAPACITY, team_size=TEAM_SIZE,
        )

def test_rejects_too_few_people_to_fill_a_and_b():
    grade_levels = {f"p{i}": 10 for i in range(4)}
    events = ["only"]
    with pytest.raises(ValueError, match="not enough people"):
        build_teams(
            grade_levels, {}, {}, dict.fromkeys(grade_levels, 1),
            {"only": 0}, events, {"only": 1}, team_size=2
        )

# ---------------------------------------------------------------------------
# Robustness across many random schools
# ---------------------------------------------------------------------------

@pytest.mark.parametrize("seed", range(8))
def test_random_schools_all_satisfy_every_criterion(seed):
    grade_levels, preferences, scores, experience = make_scenario(seed)
    roster = build_teams(
        grade_levels, preferences, scores, experience, EVENT_TIME_BLOCKS, EVENTS,
        EVENT_CAPACITY, team_size=TEAM_SIZE,
    )
    # build_teams re-verifies every criterion internally; assert the headline ones
    # here too so a regression in that checker cannot hide a broken roster.
    assert len(roster.teams["A"]) == len(roster.teams["B"]) == TEAM_SIZE
    for team in ("A", "B"):  # C is the exempt overflow team
        assert sum(
            1 for p in roster.teams[team] if grade_levels[p] == SENIOR_GRADE
        ) <= 7
    for team in TEAMS:
        for event in EVENTS:
            assert len(roster.event_assignments[team][event]) == EVENT_CAPACITY[event]
        for person, events in roster.person_events.items():
            assert len(events) <= 3
            blocks = [EVENT_TIME_BLOCKS[e] for e in events]
            assert len(blocks) == len(set(blocks))


if __name__ == "__main__":
    result = build()
    print(f"{PEOPLE_COUNT} students, {len(EVENTS)} events, "
          f"{len(set(EVENT_TIME_BLOCKS.values()))} time blocks\n")
    for team in TEAMS:
        seniors = sum(
            1 for p in result.teams[team] if GRADE_LEVELS[p] == SENIOR_GRADE
        )
        granted = sum(
            1
            for event, people in result.event_assignments[team].items()
            for person in people
            if event in PREFERENCES[person]
        )
        print(
            f"Team {team}: {len(result.teams[team]):2d} members, {seniors} seniors, "
            f"objective {result.objective[team]:6.2f}, "
            f"mean score {_mean_score_of_assignments(result, team):.3f}, "
            f"granted: {granted:2d} preferences granted"
        )

    print("\nEvent line-ups")
    for event in EVENTS:
        cells = " | ".join(
            f"{team}: {', '.join(result.event_assignments[team][event])}"
            for team in TEAMS
        )
        print(f" {event:22s} block {EVENT_TIME_BLOCKS[event]} {cells}")

    '''raise SystemExit(pytest.main([__file__, "-q"]))'''
````

````python
# Optimization algorithm to assign 2 people per event with a 3-event limit per person

def assign_people_to_events(people_list, events_list):
    # Keep track of how many events each person is assigned to
    assignment_count = {person.name: 0 for person in people_list}
    # Final mapping of event -> list of assigned people names
    event_assignments = {event: [] for event in events_list}

    # For each event, find the best candidates based on their scores
    for event in events_list:
        # Sort all people by their specific score for this event in descending order
        # person.scores is a dict where keys are event names
        potential_candidates = sorted(
            people_list,
            key=lambda p: p.scores.get(event, 0),
            reverse=True
        )

        # Attempt to assign the top 2 people who haven't reached their 3-event limit
        count = 0
        for person in potential_candidates:
            if count < EVENT_CAPACITY[event]:
                if assignment_count[person.name] < 3:
                    event_assignments[event].append(person.name)
                    assignment_count[person.name] += 1
                    count += 1
            else:
                break

    return event_assignments
````

````python
import pulp

def optimize_scioly_teams(students, EVENTS, EVENT_CAPACITY, EVENT_BLOCKS, max_team_size=15):
    """
    Optimizes team placement based on scores, event capacities, block conflicts,
    and limits on seniors per team.
    """
    # Initialize the LP Problem to Maximize total weighted score
    prob = pulp.LpProblem("SciOly_Team_Optimization", pulp.LpMaximize)

    # Teams and prioritization weights to place top students in A, mid in B, low in C
    TEAMS = ['A', 'B', 'C']
    TEAM_WEIGHTS = {'A': 10000, 'B': 100, 'C': 1}

    # --- 1. Define Variables ---
    # team_vars[i][t] = 1 if student i is on team t, else 0
    team_vars = {}
    # event_vars[i][t][e] = 1 if student i is on team t doing event e, else 0
    event_vars = {}

    for i, student in enumerate(students):
        team_vars[i] = {}
        event_vars[i] = {}
        for t in TEAMS:
            team_vars[i][t] = pulp.LpVariable(f"team_{i}_{t}", cat='Binary')
            event_vars[i][t] = {}
            for e in EVENTS:
                event_vars[i][t][e] = pulp.LpVariable(f"event_{i}_{t}_{e}", cat='Binary')

    # --- 2. Define Constraints ---
    for i, student in enumerate(students):
        # A student can be on at most one team (no crossing)
        prob += pulp.lpSum(team_vars[i][t] for t in TEAMS) <= 1, f"Max_one_team_per_student_{i}"

        for t in TEAMS:
            # If a student is on a team, they must have at least 1 event assigned to them
            prob += pulp.lpSum(event_vars[i][t][e] for e in EVENTS) >= team_vars[i][t], f"Must_have_event_{i}_{t}"

            # Each person can have at most 3 events per team
            prob += pulp.lpSum(event_vars[i][t][e] for e in EVENTS) <= 3 * team_vars[i][t], f"Max_3_events_student_{i}_team_{t}"

            for e in EVENTS:
                # A student can only do an event for a team if they are actually assigned to that team
                prob += event_vars[i][t][e] <= team_vars[i][t], f"Must_be_on_team_{t}_to_do_{e}_student_{i}"

                # If a student has no score data for an event, they shouldn't be assigned to it
                '''
                if e not in student.scores:
                    prob += event_vars[i][t][e] == 0, f"No_score_no_event_{i}_{t}_{e}"
                '''
            # Time block conflicts (Blocks 1-6 cannot overlap for the same student)
            # We skip block 7 because those are usually build/self-scheduled events
            for b in range(1, 7):
                events_in_block = [e for e in EVENTS if EVENT_BLOCKS.get(e) == b]
                prob += pulp.lpSum(event_vars[i][t][e] for e in events_in_block) <= 1, f"Block_{b}_conflict_student_{i}_team_{t}"

    for t in TEAMS:
        # Max 7 seniors per team
        prob += pulp.lpSum(team_vars[i][t] for i, s in enumerate(students) if s.grade) <= 7, f"Max_7_seniors_team_{t}"

        # Max standard team size (15 members)
        prob += pulp.lpSum(team_vars[i][t] for i in range(len(students))) <= max_team_size, f"Max_team_size_{t}"

        # Appropriate number of people per event per team
        for e in EVENTS:
            # Maximum capacity constraint (e.g., no more than 2 or 3 people)
            prob += pulp.lpSum(event_vars[i][t][e] for i in range(len(students))) <= EVENT_CAPACITY.get(e, 2), f"Capacity_team_{t}_event_{e}"

            # Require AT LEAST 1 person per event per team
            # prob += pulp.lpSum(event_vars[i][t][e] for i in range(len(students))) >= 1, f"Min_Capacity_team_{t}_event_{e}"

    # --- 3. Objective Function ---
    # Maximize the scores multiplied by the team weight.
    objective_terms = []
    for i, student in enumerate(students):
        for t in TEAMS:
            for e in EVENTS:
                if e in student.scores:
                    score = student.scores[e]
                    weight = TEAM_WEIGHTS[t]
                    objective_terms.append(event_vars[i][t][e] * score * weight)

    prob += pulp.lpSum(objective_terms), "Maximize_Weighted_Scores"

    # --- 4. Solve and Parse Results ---
    prob.solve(pulp.PULP_CBC_CMD(msg=0)) # msg=0 hides solver logs

    if pulp.LpStatus[prob.status] != 'Optimal':
        print("Could not find an optimal solution!")
        return None

    results = {t: {'members': [], 'events': {e: [] for e in EVENTS}} for t in TEAMS}
    for i, student in enumerate(students):
        for t in TEAMS:
            if pulp.value(team_vars[i][t]) == 1.0:
                results[t]['members'].append((student.name, student.grade))
                for e in EVENTS:
                    if pulp.value(event_vars[i][t][e]) == 1.0:
                        results[t]['events'][e].append(student.name)

    return results
````

````python
def extract_events_and_placements_with_nlp(text: str) -> list[dict[str, int]]:
    """
    Extracts events and placements by running a downloaded LLM entirely locally.
    """
    # Structure the few-shot prompt
    prompt = ("You are a strict data extraction assistant. Extract the events and their corresponding placements from the text. "
               "Output STRICTLY a valid JSON list of objects and absolutely nothing else. Do not use markdown formatting. "
               "Each object must have the event name (lowercase string) as the key and the placement (integer) as the value. "
               "The only possible event names are: ")
    prompt += ", ".join([event for event, field in EVENTS_MAP.items()])
    messages = [
        {
            "role": "system",
            "content": prompt
        },
        {
            "role": "user",
            "content": 'Text: "i placed 5th in chemistry lab and 7th in disease detectives"'
        },
        {
            "role": "assistant",
            "content": '[{"chemistry lab": 5}, {"disease detectives": 7}]'
        },
        {
            "role": "user",
            "content": f'Text: "{text}"'
        }
    ]

    # Format the messages into the exact prompt template TinyLlama expects
    prompt = extractor_pipeline.tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    try:
        # Run the local model
        outputs = extractor_pipeline(
            prompt,
            max_new_tokens=5000,
            temperature=0.1,
            do_sample=True,
            return_full_text=False # Ensures we only get the new generated text back
        )

        result_text = outputs[0]["generated_text"].strip()

        # Clean up the output in case the model adds markdown code blocks (```json)
        if result_text.startswith("```"):
            result_text = "\n".join(result_text.split("\n")[1:-1])

        return json.loads(result_text)

    except Exception as e:
        print(f"An error occurred: {e}")
        return []
````
